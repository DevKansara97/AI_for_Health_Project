```python
# Function to parse lesion size from report texts
def extract_lesion_size(text):
    if not isinstance(text, str) or pd.isna(text):
        return np.nan

    # Common patterns: '12 mm', '1.2 cm', '12x10 mm', 'measuring 15 mm', 'lesion of 8mm'
    text_lower = text.lower()

    # Try to find mentions of lesion, focus, nodule or size measurements
    # Pattern 1: look for dimensions in mm/cm close to words like 'lesion', 'nodule', 'index', 'pirads'
    patterns = [
        r'(?:lesion|nodule|focus|size|measuring|measures|up to|diameter)[^.\n]*?(\d+(?:\.\d+)?)\s*(?:mm|millimeter)',
        r'(?:lesion|nodule|focus|size|measuring|measures|up to|diameter)[^.\n]*?(\d+(?:\.\d+)?)\s*(?:cm|centimeter)',
        r'(\d+(?:\.\d+)?)\s*(?:mm|millimeter)\s*(?:lesion|nodule|focus|index)',
        r'(\d+(?:\.\d+)?)\s*(?:cm|centimeter)\s*(?:lesion|nodule|focus|index)'
    ]

    for pattern in patterns:
        matches = re.findall(pattern, text_lower)
        if matches:
            # Take the first matched value and convert to float mm
            val = float(matches[0])
            # If unit is cm, convert to mm
            if 'cm' in pattern or 'centimeter' in pattern:
                val *= 10.0
            return val

    return np.nan

# Create a mask for missing values
is_missing = features['lesion_max_mm'].isna()
print(f"Initial missing count: {is_missing.sum()}")

if report_col and report_col in features.columns:
    extracted_sizes = features[report_col].apply(extract_lesion_size)
    # Fill missing values with extracted sizes where possible
    features.loc[is_missing, 'lesion_max_mm'] = features.loc[is_missing, 'lesion_max_mm'].fillna(extracted_sizes)

# Add the binary indicator for cases that remain missing
features['lesion_size_missing'] = features['lesion_max_mm'].isna().astype(int)

# For any remaining missing values, we fill with 0
features['lesion_max_mm'] = features['lesion_max_mm'].fillna(0.0)

print(f"Remaining missing count after extraction: {features['lesion_size_missing'].sum()}")
print("Updated DataFrame shape:", features.shape)
```

```python
# Reasoning for keeping and imputing lab_hba1c:
# HbA1c is typically only measured in patients with known diabetes or metabolic risk factors.
# Completely dropping it would discard a potentially useful clinical subgroup indicator.
# Therefore, we use a missingness indicator (which acts as a proxy for "non-diabetic / not tested")
# combined with median imputation for the numerical column.

# 1. Create the binary missingness indicator
features['lab_hba1c_missing'] = features['lab_hba1c'].isna().astype(int)

# 2. Impute missing values with the median of the available lab_hba1c readings
hba1c_median = features['lab_hba1c'].median()
features['lab_hba1c'] = features['lab_hba1c'].fillna(hba1c_median)

print(f"Median HbA1c used for imputation: {hba1c_median}")
print("Missing values remaining in 'lab_hba1c':", features['lab_hba1c'].isna().sum())
print("DataFrame shape:", features.shape)
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Handle highly skewed variables
skewed_cols = ['psa', 'psad', 'psap', 'psa_slope_per_step', 'psa_last_over_first', 'vol']

# Apply log1p to right-skewed variables
for col in skewed_cols:
    features[f'{col}_log1p'] = np.log1p(features[col])

# For psav (contains negative values), apply a signed log transform: sign(x) * log1p(|x|)
features['psav_log1p'] = np.sign(features['psav']) * np.log1p(np.abs(features['psav']))

# Scale transformed columns
columns_to_scale = [f'{col}_log1p' for col in skewed_cols] + ['psav_log1p']
scaler = StandardScaler()
features[columns_to_scale] = scaler.fit_transform(features[columns_to_scale])

# Drop original skewed columns to avoid duplication/collinearity in modeling
features = features.drop(columns=skewed_cols + ['psav'])

# 2. Treat cspca sentinel 0.0 values as missing
features['cspca_missing'] = (features['cspca'] == 0.0).astype(int)
features['cspca'] = features['cspca'].replace(0.0, np.nan)

# Impute missing cspca values with the median of valid cspca values
cspca_median = features['cspca'].median()
features['cspca'] = features['cspca'].fillna(cspca_median)

# 3. Drop unstructured 'psa_trend_values' column as trends are summarized by structured features
if 'psa_trend_values' in features.columns:
    features = features.drop(columns=['psa_trend_values'])

print("Transformations completed!")
print("Features scaled:", columns_to_scale)
print("Sentinel 0.0 rows in 'cspca' flagged & imputed with median:", cspca_median)
print("New dataset shape:", features.shape)
```

```python
import re
import numpy as np
import pandas as pd

# Combine text columns to search comprehensively across all clinical narratives
text_cols_to_process = ['note_history', 'previous_notes', 'radiology_report', 'note_chief_complaint', 'note_physical_examination']
combined_text = features[text_cols_to_process].fillna('').agg(' '.join, axis=1).str.lower()

# 1. Extract Prior ISUP / Gleason Grade
# Looks for 'isup 1-5', 'isup grade 1-5', 'gleason 3+3', 'gleason 3+4', 'gleason 4+3', etc.
def extract_prior_isup(text):
    isup_match = re.search(r'isup\s*(?:grade\s*)?([1-5])', text)
    if isup_match:
        return int(isup_match.group(1))

    gleason_match = re.search(r'gleason\s*(?:score\s*)?(\d)\s*\+\s*(\d)', text)
    if gleason_match:
        g1, g2 = int(gleason_match.group(1)), int(gleason_match.group(2))
        total = g1 + g2
        if total <= 6:
            return 1
        elif g1 == 3 and g2 == 4:
            return 2
        elif g1 == 4 and g2 == 3:
            return 3
        elif total == 8:
            return 4
        elif total >= 9:
            return 5
    return 0

features['prior_isup_grade'] = combined_text.apply(extract_prior_isup)
features['has_prior_isup'] = (features['prior_isup_grade'] > 0).astype(int)

# 2. Extract Active Surveillance Status
as_keywords = ['active surveillance', ' surveillance', 'watching waiting', 'watchful waiting', 'on as']
features['active_surveillance'] = combined_text.apply(lambda x: 1 if any(kw in x for kw in as_keywords) else 0)

# 3. Extract Lesion Growth Compared with Previous MRI
growth_keywords = ['increase in size', 'increased in size', 'growth', 'larger', 'progression', 'growing', 'interval increase', 'interval growth']
features['lesion_growth'] = combined_text.apply(lambda x: 1 if any(kw in x for kw in growth_keywords) else 0)

# Drop raw unstructured text columns to complete cleaning
features = features.drop(columns=text_cols_to_process)

print("Free-text extraction complete!")
print(f"Records with prior ISUP grade: {features['has_prior_isup'].sum()}")
print(f"Records on Active Surveillance: {features['active_surveillance'].sum()}")
print(f"Records indicating lesion growth: {features['lesion_growth'].sum()}")
print("New dataset shape:", features.shape)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# 1. Filter out the 104 unlabeled rows
labeled_df = features[features['label'].notna()].copy()
labeled_df['label'] = labeled_df['label'].astype(int)

# 2. Define target (y) and features (X)
y = labeled_df['label']

# Exclude identifiers, targets, and any post-decision audit columns
exclude_cols = ['case_id', 'label']
X = labeled_df.drop(columns=exclude_cols)

# Identify numerical and categorical features for pipeline processing
# Note: most categorical features have already been one-hot encoded, but we can verify types
num_cols = X.select_dtypes(include=[np.number]).columns.tolist()

print(f"Labeled dataset size: {X.shape}")
print(f"Number of features: {len(num_cols)}")
print(f"Class distribution: {np.bincount(y)} (ratio: {np.bincount(y)[1]/len(y):.2f})")
```

```python
# 3. Define the preprocessing pipeline
# We fit the imputers and scalers *inside* each fold to avoid data leakage
preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), num_cols)
    ]
)

# 4. Construct the modeling pipeline (using a robust Logistic Regression baseline)
model_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced'))
])

# 5. Set up Repeated Stratified K-Fold (5 repeats, 10 folds)
rskf = RepeatedStratifiedKFold(n_splits=10, n_repeats=5, random_state=42)

# 6. Evaluate using cross_validate
scoring = ['roc_auc', 'balanced_accuracy']
scores = cross_validate(model_pipeline, X, y, cv=rskf, scoring=scoring, n_jobs=-1)

# 7. Print baseline results
print(f"--- Baseline Logistic Regression Performance (5x10 RSKF) ---")
print(f"Mean ROC AUC: {scores['test_roc_auc'].mean():.4f} +/- {scores['test_roc_auc'].std():.4f}")
print(f"Mean Balanced Accuracy: {scores['test_balanced_accuracy'].mean():.4f} +/- {scores['test_balanced_accuracy'].std():.4f}")
```
